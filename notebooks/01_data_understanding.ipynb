# Data Understanding

El objetivo de este notebook es comprender la estructura del dataset Superstore, identificar las variables disponibles y detectar posibles problemas de calidad de datos antes de iniciar la fase de limpieza.

Carga de datos

```python
import pandas as pd

df = pd.read_csv("../data/raw/Sample - Superstore.csv", encoding="latin1")

df.head()
```

El dataset se carga desde la carpeta `data/raw`.  
Se utiliza `encoding="latin1"` porque el archivo presenta una codificación diferente a UTF-8.

Tamaño del dataset

```python
df.shape
```

El dataset contiene 9.994 registros y 21 variables.

Información general

```python
df.info()
```

La mayoría de variables son categóricas, aunque también existen variables numéricas como Sales, Quantity, Discount y Profit.

No se observan valores nulos en las columnas del dataset.

Estadísticas descriptivas

```python
df.describe()
```

Las estadísticas descriptivas permiten observar el comportamiento general de las variables numéricas.

Se observa que Profit incluye valores negativos, lo que indica que algunos pedidos generan pérdidas.

Revisión de duplicados

```python
df.duplicated().sum()
```

No se encontraron registros duplicados en el dataset.

Revisión de categorías principales

```python
df["Category"].value_counts()
```

```python
df["Region"].value_counts()
```

```python
df["Segment"].value_counts()
```

El dataset contiene tres categorías principales de productos: Office Supplies, Furniture y Technology.

También incluye cuatro regiones de venta y tres segmentos de clientes.

Valores únicos

```python
df.nunique()
```

El dataset contiene 793 clientes únicos, 1.850 productos únicos y 5.009 pedidos únicos.

Esto permite realizar análisis desde diferentes perspectivas: clientes, productos, pedidos y regiones.

Ventas y beneficios totales

```python
df["Sales"].sum()
```

```python
df["Profit"].sum()
```

El dataset registra unas ventas totales aproximadas de 2.297.200 dólares y un beneficio total aproximado de 286.397 dólares.

Primer análisis por categoría

```python
df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
```

```python
df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
```

Technology es la categoría con mayores ventas y beneficios.

Furniture presenta un volumen de ventas elevado, pero su beneficio es considerablemente menor en comparación con Technology y Office Supplies.

## Conclusión

El dataset Superstore contiene información suficiente para realizar un análisis de negocio completo.

No presenta valores nulos ni duplicados, por lo que se encuentra en buen estado para continuar con la fase de limpieza y preparación.

Los primeros análisis muestran que Technology es la categoría con mejores resultados, mientras que Furniture podría presentar problemas de rentabilidad que serán investigados en el análisis de negocio.
