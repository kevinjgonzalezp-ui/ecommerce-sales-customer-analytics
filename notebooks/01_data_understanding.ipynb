# 1. Comprensión Inicial del Dataset

El dataset Superstore contiene 9.994 registros y 21 variables relacionadas con ventas, clientes, productos y regiones.

Tras una primera exploración, se observa que no existen valores nulos en ninguna de las columnas, lo que facilita el proceso de limpieza y preparación de datos.

Además, el conjunto de datos incluye información sobre ventas, beneficios, descuentos, cantidades vendidas, clientes y ubicaciones geográficas, lo que lo convierte en una fuente adecuada para realizar análisis de negocio y desarrollar un dashboard interactivo en Power BI.

## Principales hallazgos iniciales

Tras la exploración inicial del dataset se han identificado las siguientes características:

- El conjunto de datos contiene 9.994 transacciones y 21 variables.
- No existen valores nulos ni registros duplicados.
- La categoría con mayor número de ventas registradas es Office Supplies, seguida de Furniture y Technology.
- El negocio opera en cuatro regiones: West, East, Central y South.
- El segmento de clientes más representado es Consumer, seguido por Corporate y Home Office.

## Análisis inicial de ventas y beneficios

Las ventas totales del negocio ascienden a 2.297.200 dólares, generando un beneficio total de 286.397 dólares.

Al analizar las categorías principales, se observa que Technology es la categoría con mayores ventas y también la más rentable.

Por otro lado, Furniture presenta un comportamiento interesante: aunque genera un volumen de ventas similar al de Technology, su beneficio es considerablemente inferior. Esto sugiere la existencia de factores que podrían estar afectando negativamente a su rentabilidad y que serán analizados en etapas posteriores del proyecto.

Office Supplies se sitúa como la segunda categoría más rentable a pesar de no ser la que registra el mayor volumen de ventas.

```python
df.groupby("Region")[["Sales","Profit"]].sum().sort_values("Sales", ascending=False)
```

```python
df.groupby("Segment")[["Sales","Profit"]].sum().sort_values("Sales", ascending=False)
```

## Análisis por región y segmento de clientes

El análisis regional muestra que West es la región con mayor volumen de ventas y beneficio, seguida de East.

Sin embargo, se observa un comportamiento interesante entre las regiones Central y South. Aunque Central registra un mayor volumen de ventas, South genera un beneficio superior, lo que podría indicar una mejor rentabilidad o una estrategia comercial más eficiente.

Respecto a los segmentos de clientes, Consumer representa el grupo más importante para la compañía, concentrando el mayor volumen de ventas y beneficios. Corporate ocupa la segunda posición, mientras que Home Office representa la menor participación dentro del negocio.
