# 2. Limpieza y preparación de datos

En esta fase se realizan las transformaciones necesarias para preparar el dataset para el análisis posterior y la construcción del dashboard en Power BI.

```python
import pandas as pd

df = pd.read_csv("../data/raw/Sample - Superstore.csv", encoding="latin1")

df.head()
```

## Conversión de variables de fecha

Las columnas de fecha se convierten al formato datetime para facilitar el análisis temporal.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

df[["Order Date","Ship Date"]].dtypes
```

## Creación de variables temporales

Se crea la variable Year para analizar la evolución de las ventas a lo largo del tiempo.

```python
df["Year"] = df["Order Date"].dt.year

df["Year"].value_counts()
```

## Verificación final

Se comprueba que el dataset se encuentra correctamente preparado para la fase de análisis de negocio.

```python
df.info()
```

```python
df.head()
```

## Exportación del dataset limpio

Se guarda el dataset preparado en la carpeta `data/cleaned` para utilizarlo posteriormente en SQL y Power BI.

```python
df.to_csv(
    "../data/cleaned/superstore_clean.csv",
    index=False
)
```
